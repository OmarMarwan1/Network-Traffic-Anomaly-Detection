## Now for the big thing 
we will try to classify for all the days the UNB has to offer in that file 
we will:
1. put all the files in one folder
2. Read the file
3. clean the colums name
4. drop duplicates
5. drop the shortcut feature if exists
6. Handle infinite values and NaNs
7. Merge all cleaned chunks into one master dataset

```python
import os
def merge_files(FOLDER_PATH):
    folder_path = FOLDER_PATH
    csv_files = [f for f in os.listdir(folder_path) if f.endswith('.csv')]
    dataframes = []
    for file in csv_files:
        print("processing files")
        df = pd.read_csv(os.path.join(folder_path , file))
        df.columns = df.columns.str.strip()
        df.drop_duplicates(inplace= True)
        if "Destination Port" in df.columns:
            df.drop("Destination Port" , axis = 1 , inplace = True)
        df.replace([np.inf , -np.inf] , np.nan , inplace = True)
        df.dropna(inplace= True)
        dataframes.append(df)
    print("Merging all days into one dataset...")
    master_df = pd.concat(dataframes , ignore_index = True)
    del dataframes
    print(f"final dataset shape :{master_df.shape}")
    return master_df
```

```python
from sklearn.preprocessing import LabelEncoder 

def preprocess(master_df):
    # Use "Label" (capital L) unless you explicitly lowercased your columns earlier
    x = master_df.drop("Label", axis=1)
    y = master_df["Label"]
    
    encoder = LabelEncoder()
    y_encoded = encoder.fit_transform(y)
    
    label_mapping = dict(zip(encoder.transform(encoder.classes_), encoder.classes_))
    print("Attack Label Mapping:")
    for num, name in label_mapping.items():
        print(f"{num} : {name}")
        
    # You must return x, y_encoded, AND the encoder so you can use them later
    return x, y_encoded, encoder
```

```python
full_data_set = merge_files(r"D:\sample_project_1\network_traffic")
```

```python
preprocess(full_data_set)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report

#run your function to extract the features, numerical labels, and the encoder
X, y_encoded, label_encoder = preprocess(full_data_set)

#split the massive dataset
X_train, X_test, y_train, y_test = train_test_split(X, y_encoded, test_size=0.2, random_state=42)

#train on all 12 CPU threads
print("Training the Multi-Class Model...")
model = RandomForestClassifier(n_jobs=-1, random_state=42 , class_weight = "balanced")
model.fit(X_train, y_train)

#predict and evaluate
preds = model.predict(X_test)

#use label_encoder.classes_ to print the actual attack names instead of just numbers
print(classification_report(y_test, preds, target_names=label_encoder.classes_, digits=4))
```

```python
import os

def merge_files1(FOLDER_PATH):
    csv_files = [f for f in os.listdir(FOLDER_PATH) if f.endswith('.csv')]
    dataframes = []

    for file in csv_files:
        print(f"Processing {file}")

        df = pd.read_csv(os.path.join(FOLDER_PATH, file))

        df.columns = df.columns.str.strip()

        df.drop_duplicates(inplace=True)

        if "Destination Port" in df.columns:
            df.drop("Destination Port", axis=1, inplace=True)

        df.replace([np.inf, -np.inf], np.nan, inplace=True)
        df.dropna(inplace=True)

        # Keep track of which day/file this data came from
        df["day"] = file

        dataframes.append(df)

    print("Merging all days into one dataset...")

    master_df = pd.concat(dataframes, ignore_index=True)

    del dataframes

    print(f"Final dataset shape: {master_df.shape}")

    return master_df
```

```python
import pandas as pd
import numpy as np
import os

def load_dataset(folder_path):

    dataframes = []

    for file in os.listdir(folder_path):

        if not file.endswith(".csv"):
            continue

        print("Processing:", file)

        df = pd.read_csv(
            os.path.join(folder_path, file),
            low_memory=False
        )

        df.columns = df.columns.str.strip()

        df.drop_duplicates(inplace=True)

        if "Destination Port" in df.columns:
            df.drop(columns=["Destination Port"], inplace=True)

        df.replace([np.inf, -np.inf], np.nan, inplace=True)

        df.dropna(inplace=True)

        df["day"] = file

        dataframes.append(df)

    print("Merging...")

    master_df = pd.concat(dataframes, ignore_index=True)

    return master_df
```
